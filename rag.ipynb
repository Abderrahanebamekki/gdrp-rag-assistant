```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_ollama import OllamaEmbeddings, ChatOllama
from langchain_chroma import Chroma
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough
```

```python
# ---- Step 1: Load the PDF ----
loader = PyPDFLoader("cv.pdf")   
documents = loader.load()
```

```python
# ---- Step 2: Split into chunks ----
splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200
)
chunks = splitter.split_documents(documents)
```

```python
# ---- Step 3: Embed chunks + store in Chroma ----
embeddings = OllamaEmbeddings(model="nomic-embed-text")

vectorstore = Chroma.from_documents(
    documents=chunks,
    embedding=embeddings,
    persist_directory="./chroma_db"   # saved to disk, so you don't re-embed every run
)

retriever = vectorstore.as_retriever(search_kwargs={"k": 4})  # top 4 relevant chunks
```

```python
# ---- Step 4: Build the prompt ----
template = ChatPromptTemplate.from_messages([
    ("system", """You are a helpful assistant answering questions based only on the provided context.
If the answer isn't in the context, say you don't know — don't make things up.

Context:{context}"""),

     ("human", "{question}")
])

```

```python
# ---- Step 5: LLM ----
llm = ChatOllama(model="llama3.2:3b", temperature=0.0)

# ---- Step 6: Build the full RAG chain ----
def format_docs(docs):
    return "\n\n".join(doc.page_content for doc in docs)

rag_chain = (
    {"context": retriever | format_docs, "question": RunnablePassthrough()}
    | template
    | llm
    | StrOutputParser()
)
```

```python

# ---- Step 7: Ask questions ----
if __name__ == "__main__":
    while True:
        question = input("\nAsk a question about the PDF (or 'exit'): ")
        if question.lower() == "exit":
            break
        answer = rag_chain.invoke(question)
        print(f"\nAnswer: {answer}")
```
